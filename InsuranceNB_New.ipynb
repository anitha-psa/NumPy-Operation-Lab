```python
# Step 1 Import and Load Dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import mean_squared_error, r2_score

# Load dataset
df = pd.read_csv("insurance.csv")  # Replace with your actual path
df.head()
```

```python
# Step 2 Exploratory Data Analysis

df.info()
df.describe()

# Check for missing values
print(df.isnull().sum())

# Correlation heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

```python
# Step 3 Data Preprocessing

# Step 3.1 Convert Categorical Variables
df_encoded = df.copy()

# One-hot encoding for categorical columns
df_encoded = pd.get_dummies(df_encoded, columns=['sex', 'smoker', 'region'], drop_first=True)
df_encoded.head()
```

```python
# Step 3.2 Split features and Taget
X = df_encoded.drop('charges', axis=1)
y = df_encoded['charges']
```

```python
# Step 4 Train-Test Split (70-30)

X_train, X_test, y_train, y_test = train_test_split(    X, y, test_size=0.3, random_state=42)
```

```python
# Step 5: Train Linear Regression Model

model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# Step 6 Evaluate the Model

y_pred = model.predict(X_test)

# Performance metrics
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("Mean Squared Error (MSE):", mse)
print("Root Mean Squared Error (RMSE):", rmse)
print("R² Score:", r2)
```

```python
# Optional VS Predicted plot

plt.scatter(y_test, y_pred, alpha=0.5)
plt.xlabel("Actual Charges")
plt.ylabel("Predicted Charges")
plt.title("Actual vs Predicted Insurance Charges")
plt.show()
```

## Observations:

- The linear regression model shows a decent fit, though there's room for improvement.
- RMSE gives us an idea of average prediction error.
- R² score indicates how much variance is explained by the model (closer to 1 is better).
- Possible improvements:
  - Feature engineering (e.g., BMI interaction terms)
  - Try regularized models (Ridge, Lasso)
  - Outlier removal
